```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Визначення системи
def system(state, t):
    x1, x2 = state
    dx1dt = 4 * x1 * (x2 - 1)
    dx2dt = x2 * (x1 + x1**2)
    return [dx1dt, dx2dt]

# Підготовка сітки для векторного поля (Streamplot)
x = np.linspace(-3, 3, 100)
y = np.linspace(-1, 4, 100)
X, Y = np.meshgrid(x, y)

# Обчислення векторів швидкості в кожній точці сітки
U = 4 * X * (Y - 1)
V = Y * (X + X**2)

# Налаштування графіку
plt.figure(figsize=(10, 8))

# лінії потоку (глобальний портрет)
plt.streamplot(X, Y, U, V, density=1.5, color='lightgray', linewidth=0.8)

# ізокліни
plt.axvline(x=-1, color='red', linestyle='--', label='ẋ1=0 (x1=-1)')
plt.axhline(y=1, color='blue', linestyle='--', label='ẋ2=0 (x2=1)')
plt.axvline(x=0, color='green', linewidth=2, label='x1=0')

plt.plot(-1, 1, 'ro', markersize=10, label='Saddle Point (-1, 1)')

plt.title("Фазовий портрет системи")
plt.xlabel("x1")
plt.ylabel("x2")
plt.xlim(-3, 3)
plt.ylim(-1, 4)
plt.legend()
plt.grid()
plt.show()
```
