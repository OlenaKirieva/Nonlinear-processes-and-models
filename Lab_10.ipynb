```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Визначення системи рівнянь
def system(t, z):
    x1, x2 = z
    dx1 = x2 - x1**3 + x1 * x2**2
    dx2 = -x1 - x1 * x2**2
    return [dx1, dx2]

# Параметри часу
t_span = [0, 100]  # Час моделювання
z0 = [0.1, 0.1]    # Початкова точка (близько до 0)

# Розв'язок системи
sol = solve_ivp(system, t_span, z0, rtol=1e-9, atol=1e-9, max_step=0.1)

# Малюємо графіки
plt.figure(figsize=(12, 5))

# Графік оригінальної системи
# plt.subplot(1, 2, 1)
plt.plot(sol.y[0], sol.y[1])
plt.plot(0, 0, 'ro', label='Початок (0,0)')
plt.title("Фазовий портрет")
plt.xlabel("x1")
plt.ylabel("x2")
plt.grid(True)
plt.axis('equal');
```
