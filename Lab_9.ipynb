```python
import numpy as np
import matplotlib.pyplot as plt

# Визначення системи
def system(state, t):
    x1, x2 = state
    term_exp = np.exp(x1 + x2)
    dx1dt = term_exp
    dx2dt = x1 + x2
    return [dx1dt, dx2dt]

# Підготовка сітки
x_range = np.linspace(-3, 3, 20)
y_range = np.linspace(-3, 3, 20)
X, Y = np.meshgrid(x_range, y_range)

# Векторне поле
U = np.exp(X + Y)
V = X + Y

# Нормалізація стрілок для streamplot
speed = np.sqrt(U**2 + V**2)
U_norm = U / speed
V_norm = V / speed

plt.figure(figsize=(8, 6))

# Фазовий портрет
st = plt.streamplot(X, Y, U, V, color=speed, cmap='autumn', density=1.2, linewidth=1)

plt.title("Фазовий портрет системи")
plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.xlim(-3, 3)
plt.ylim(-3, 3)
plt.grid()
plt.show()
```
